```python
import whisper
import os
import re
import json
```

```python
input_path = "/content/Exemple2 RGPD.WAV"
```

```python
# 5. NETTOYAGE AUDIO AVEC FFMPEG
cleaned_path = "/content/cleaned_audio.wav"
os.system(f"ffmpeg -i '{input_path}' -ar 16000 -ac 1 -af 'afftdn' '{cleaned_path}'")
```

```python
# 6. CHARGEMENT DU MODÈLE WHISPER (large-v3 pour qualité maximale)
model = whisper.load_model("large-v3")
```

```python
# 7. TRANSCRIPTION EN FRANÇAIS
result = model.transcribe(cleaned_path, language="fr", verbose=False)
```

```python
# 8. AFFICHAGE DU TEXTE FINAL
print("🎧 Transcription :")
print("="*80)
print(result["text"])
```
